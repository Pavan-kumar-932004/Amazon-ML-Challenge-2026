```python
import kagglehub

path = kagglehub.dataset_download("satwiksps/amazon-ml-challenge-2026")

print("Path to dataset files:", path)
```

```python
import shutil
import os

content_path = 'D:/Amazon-ML-Challenge-2026'

os.makedirs(content_path, exist_ok=True)

for item in os.listdir(path):
    source_item = os.path.join(path, item)
    destination_item = os.path.join(content_path, item)
    if os.path.isdir(source_item):
        shutil.copytree(source_item, destination_item, dirs_exist_ok=True)
    else:
        shutil.copy2(source_item, destination_item)

print(f"Dataset successfully copied from {path} to {content_path}")
print(f"You can now access your dataset files in: {content_path}")
```
